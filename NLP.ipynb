```python
corpus = "This is a paragraph. This conists of lines ending with , . and ; . This consists of various words"
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import sent_tokenize
doc = sent_tokenize(corpus)
doc
```

```python
for i in doc:
  print(i)
```

```python
from nltk.tokenize import word_tokenize
words = word_tokenize(corpus)
words
```

```python
from nltk.tokenize import TreebankWordTokenizer
tokenizer = TreebankWordTokenizer()
tokens = tokenizer.tokenize(corpus)
print(tokens)
```

```python
text = "Dr. Smith is here. He is nice."
words_1 = word_tokenize(text)
tokens_1 = tokenizer.tokenize(text)
print(words_1)
print(tokens_1)
```
